# Tópicos de Problemas de Otimização, utilizando programação Linear e Bibliotecas Python

  Problema baseado no livro de Luzia Kazuko Yoshida.

1. Uma pequena empresa produz antrigos $A_{1}$ e $A_{2}$, que são vendidos a 200 u.m e 300 u.m , respectivamente.
2. Na produção dos produtos, são utilizados as matérias-primas $P_{1}; P_{2}; P_{3}$ , que são gastas da seguinte forma:



* 2 unidades de $
P_{1}$ para fabricar uma unidade de $A_{1}$
* 4 uniddes de $P_{2}$ para uma unidade de $A_{1}$

* 1 unidade de $P_{1}$ para fabricar 1 unidade de $A_{2}$
* 1 unidade de $P_{3}$ para fabricar 1 unidade de $A_{2}$

## Função a ser maximizada
$f(x_{1}; x_{2} )= 200*x_{1} + 300*x_{2}$


Onde, $x_{1};x_{1}$ são as quantidades respectivasd de produção para o produto A1 e A2.



## Restrições

Por razões econômicas, as matérias-primas $P_{1}; P_{2}; P_{3}$ estão disponíveis nas seguintes quantidades:

1. $P_{1} \le 20 $;
2. $P_{2} \le 32 $;
3. $P_{3} \le 10 $.

#### Uso de matéria-prima
|    | P1 | P2 | P3 | x
|----|:--:|:--:|:--:|:--:|
| A1 |  2 |  4 |  0 | $x_{1}$|
| A2 |  1 |  0 |  1 | $x_{2}$|
|tot | 20 | 32 |  10|        |


### Restrições de matéria prima

* $ 2*x_{1} + x_{2} \le 20 $ restrição da matéria prima P1
* $ 4*x_{1} \le 32 $         restrição da matéria prima P2
* $ x_{2} \le 10 $           restrição da matéria prima P3


  Quais as quantidades dos produtos A1  e A2 que devem ser produzidas para que a receita bruta seja a maior possível?

```python
import numpy as np
from scipy.optimize import linprog

# Função objetivo: Maximizar f(x1, x2) = 200*x1 + 300*x2
# Como o linprog minimiza, usamos coeficientes negativos: -200 e -300
c = [-200, -300]

# Coeficientes das restrições de desigualdade (A_ub * x <= b_ub)
# 2*x1 + 1*x2 <= 20 (Matéria-prima P1)
# 4*x1 + 0*x2 <= 32 (Matéria-prima P2)
# 0*x1 + 1*x2 <= 10 (Matéria-prima P3)
A = [
    [2, 1],
    [4, 0],
    [0, 1]
]

b = [20, 32, 10]

# Limites das variáveis (x1 >= 0, x2 >= 0)
x0_bounds = (0, None)
x1_bounds = (0, None)

# Resolver o problema de programação linear
res = linprog(c, A_ub=A, b_ub=b, bounds=[x0_bounds, x1_bounds], method='highs')

# Apresentar os resultados
if res.success:
    print("Otimização realizada com sucesso!")
    print(f"Quantidade do produto A1 (x1): {res.x[0]:.2f} unidades")
    print(f"Quantidade do produto A2 (x2): {res.x[1]:.2f} unidades")
    print(f"Receita Bruta Máxima: R$ {-res.fun:.2f}")
else:
    print("Não foi possível encontrar uma solução ótima:", res.message)
```

O grande segredo em programação linear é modelar o problema para aplicá-lo no Python. O linprog é o Solver mais antigo, mas é muito mais intuitivo de usar do que outras bibliotecas.

# PROBLEMAS DE TRANSPORTE: Um exemplo simples
O Problema de Transporte é um caso clássico de Programação Linear, no qual o objetivo é determinar a quantidade a ser enviada de múltiplos armazéns (origens) para múltiplos centros de consumo (destinos) de forma a minimizar o custo total de frete, respeitando a oferta disponível nas origens e a demanda dos destinos.

Como o total ofertado é igual ao total consumido ($\sum a_i = \sum b_j$), o problema é considerado balanceado, e todas as restrições de restrição de capacidade/demanda tornam-se restrições de igualdade.

1. Formulação Matemática
Variáveis de decisão:

$x_{i,j}$ = quantidade transportada do armazém $i$ para o centro de consumo $j$.

Função Objetivo (Minimização):$$\min Z = \sum_{i=1}^{m} \sum_{j=1}^{n} c_{i,j} x_{i,j}$$

Restrições:Oferta (Armazéns): O total enviado de cada armazém $i$ para todos os destinos deve ser igual à sua oferta $a_i$.$$\sum_{j=1}^{n} x_{i,j} = a_i \quad (\text{para } i = 1, \dots, m)$$Demanda (Centros de Consumo): O total recebido em cada destino $j$ de todos os armazéns deve ser igual à sua demanda $b_j$.$$\sum_{i=1}^{m} x_{i,j} = b_j \quad (\text{para } j = 1, \dots, n)$$Não-negatividade:$$x_{i,j} \ge 0$$

2. Implementação com scipy.optimize.linprog

Em PythonA função linprog do SciPy resolve problemas no formato padrão matricial: $\min c^T x$ sujeito a $A_{eq} x = b_{eq}$.

Para usar o linprog, precisamos achatar (flatten) a matriz de variáveis $x_{i,j}$ em um vetor unidimensional de tamanho $m \times n$.

```python
import numpy as np
from scipy.optimize import linprog

# 1. Dados do Problema
# Custos unitários c[i, j] (Matriz m x n)
# Ex: 2 armazéns e 3 destinos
custos = np.array([
    [2, 4, 5],
    [3, 2, 1]
])
m, n = custos.shape

# Ofertas dos armazéns (a) e Demandas dos destinos (b)
oferta = np.array([100, 150])  # Total = 250
demanda = np.array([80, 90, 80])  # Total = 250 (Balanceado)

# 2. Vetor de Custos (c)
# O linprog espera um vetor 1D. Achatar a matriz de custos resolve isso.
c = custos.flatten()

# 3. Construção das Matrizes de Restrição de Igualdade (A_eq e b_eq)
# O vetor de variáveis x tem tamanho m * n.

A_eq_list = []
b_eq_list = []

# Restrições de Oferta (Linhas)
for i in range(m):
    restricao = np.zeros((m, n))
    restricao[i, :] = 1  # Soma todas as colunas para a linha i
    A_eq_list.append(restricao.flatten())
    b_eq_list.append(oferta[i])

# Restrições de Demanda (Colunas)
for j in range(n):
    restricao = np.zeros((m, n))
    restricao[:, j] = 1  # Soma todas as linhas para a coluna j
    A_eq_list.append(restricao.flatten())
    b_eq_list.append(demanda[j])

A_eq = np.array(A_eq_list)
b_eq = np.array(b_eq_list)

# 4. Limites de Não-negatividade (x >= 0)
# Por padrão, o linprog assume x >= 0, mas podemos explicitar bounds=(0, None)
bounds = [(0, None)] * (m * n)

# 5. Execução do Otimizador
resultado = linprog(c, A_eq=A_eq, b_eq=b_eq, bounds=bounds, method='highs')

# 6. Exibição dos Resultados
if resultado.success:
    print("Solução Ótima Encontrada com Sucesso!\n")
    print(f"Custo Total Mínimo de Transporte: R$ {resultado.fun:.2f}\n")

    # Remodelar o vetor solução de volta para a matriz m x n
    x_otimo = resultado.x.reshape((m, n))
    print("Matriz de Quantidades Transportadas (x_ij):")
    print(x_otimo)
else:
    print("Não foi possível encontrar uma solução ótima:", resultado.message)
```
